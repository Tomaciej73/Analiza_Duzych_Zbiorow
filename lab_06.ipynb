Zadanie 1

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, desc, sum, count

spark = SparkSession.builder \
    .master("local[2]") \
    .appName("Recipe Reviews Analysis") \
    .getOrCreate()

df_reviews = spark.read.csv('./data/Recipe Reviews and User Feedback Dataset.csv', header=True, inferSchema=True)

df_reviews = df_reviews.withColumnRenamed("_c0", "id")

df_reviews.select("reply_count") \
    .orderBy(desc("reply_count")) \
    .show(10)

df_reviews.groupBy("recipe_code") \
    .agg(sum("best_score").alias("total_best_score")) \
    .orderBy(desc("total_best_score")) \
    .show(10)

df_reviews.groupBy("recipe_code") \
    .agg(count("id").alias("total_comments")) \
    .orderBy(desc("total_comments")) \
    .show(10)

df_reviews.groupBy("stars") \
    .count() \
    .orderBy(desc("count")) \
    .show()
```

Zadanie 2

```python
df_employee = spark.read.csv('employee.csv', header=True, inferSchema=True)
df_employee.printSchema()
df_employee.show(10)
```

Zadanie 3

```python
import time
df_employee = df_employee.withColumn("salary", col("salary").cast("double"))
start_time = time.time()
count_high_salary = df_employee.filter(col("salary") > 10000).count()
end_time = time.time()
print(f"Liczba wierszy z salary > 10000: {count_high_salary}")
print(f"Czas wykonania operacji: {end_time - start_time:.2f} sekundy")
```

Zadanie 4

```python
from pyspark.ml.feature import Bucketizer
splits = [10, 20, 30, 40, 50, 60, 70]
bucketizer = Bucketizer(splits=splits, inputCol="age", outputCol="age_bucket")
df_employee_buckets = bucketizer.transform(df_employee)
df_employee_buckets.select("age", "age_bucket").show(20)
df_employee_buckets.groupBy("age_bucket") \
    .count() \
    .orderBy("age_bucket") \
    .show()
```
